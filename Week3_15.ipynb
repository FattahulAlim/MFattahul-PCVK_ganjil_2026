# Transformasi Linear Brightness

Formula : *g(x,y) = f(x,y) + b*

*g(x,y)* adalah nilai pixel setelah transformasi, *f(x,y)* adalah nilai pixel asli, b adalah nilai brightness

```python
import cv2 as cv
import numpy as np
from google.colab.patches import cv2_imshow
import matplotlib.pyplot as plt
```

```python
print(' Mengubah tingkat kecerahan citra ')
print('----------------------------------')
try:
  brightness = int(input('Masukkan nilai kecerahan: '))
except ValueError:
  print('Error, not a number')

original = cv.imread('/content/drive/MyDrive/PCVK-MFattahul/week-3/ktm1a.jpg')
brightness_image = np.zeros(original.shape, original.dtype)

#akses per piksel
#np.clip digunakan untuk melakukan truncate pixel (nilai dibatasi antara 0-255)
for y in range(original.shape[0]):
  for x in range (original.shape[1]):
    for c in range(original.shape[2]):
      brightness_image[y,x,c] = np.clip(original[y,x,c] + brightness, 0, 255)

#cara simple tanpa for loop
#brightness_image = cv.convertScaleAbs(original, beta=brightness)

# final_frame = cv.hconcat([original, brightness_image])
# cv2_imshow(final_frame)

plt.figure(figsize=(12, 5))

plt.subplot(1,2,1)
plt.imshow(cv.cvtColor(original, cv.COLOR_BGR2RGB))
plt.title('244107020018-Original')
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(cv.cvtColor(brightness_image, cv.COLOR_BGR2RGB))
plt.title('244107020018-after brightness')
plt.axis('off')

plt.show()
```

Sehingga, keunggulan Logarithmic Brightness pada kasus ini hanya disebabkan oleh kemampuannya mengangkat kecerahan piksel gelap secara jauh lebih signifikan dibandingkan Linear Brightness, bukan karena perbedaan tingkat clipping di antara keduanya.

```python
#Tugas Praktikum 8 Gaussian Noise

import glob
from math import log10, sqrt

original_galaxy = cv.imread('/content/drive/MyDrive/PCVK-MFattahul/week-3//galaxy.jpg')

cv_img = []
for img_path in glob.glob('/content/drive/MyDrive/PCVK-MFattahul/week-3/noises/*.jpg'):
 img = cv.imread(img_path)
 cv_img.append(img)

print(f'Jumlah citra noise yang terbaca: {len(cv_img)}')

def average_denoise(images, n):
    # ambil n citra pertama, jumlahkan, terus bagi n
    subset = images[:n]

    # convert ke float sebelum dijumlah, agar tidak overflow (uint8 maksimal di 255)
    total = np.zeros(subset[0].shape, dtype=np.float64)
    for img in subset:
        total += img.astype(np.float64)

    average = total / n
    average = np.clip(average, 0, 255)
    return average.astype(np.uint8)

def PSNR(img1, img2):
    img1 = img1.astype(np.float64)
    img2 = img2.astype(np.float64)
    mse = np.mean((img1 - img2) ** 2)
    if mse == 0:
        return float('inf')
    return 20 * log10(255.0 / sqrt(mse))

jumlah_list = [10, 20, 40, 80, 100]

for n in jumlah_list:
    hasil = average_denoise(cv_img, n)
    psnr_value = PSNR(original_galaxy, hasil)

    print(f'Jumlah citra: {n} | PSNR: {psnr_value:.2f} dB')
    cv2_imshow(hasil)
```

### Alasan Pemilihan:
Foto pemandangan malam hari memiliki karakteristik sebaran piksel yang timpang — sebagian besar area gelap total, namun terdapat beberapa detail (seperti kertas bertuliskan NIM dan tanggal) yang sebenarnya memiliki informasi cukup baik namun tidak terlihat karena kurangnya pencahayaan. Kedua metode ini dipilih karena sama-sama bersifat non-linear, yaitu mampu menaikkan kecerahan piksel gelap secara lebih signifikan dibandingkan piksel yang sudah terang. Hal ini penting agar detail pada area gelap dapat terangkat tanpa membuat area yang sudah bercahaya (misalnya langit atau sumber cahaya) langsung mengalami clipping menjadi putih polos, seperti risiko yang terjadi jika menggunakan Linear Brightness biasa.

### Menentukan Paremeter Terbaik:
Setelah mencoba beberapa nilai, parameter terbaik yang saya peroleh adalah Gamma = 2. Pada nilai ini, detail pada area gelap (termasuk tulisan NIM dan tanggal pada kertas) sudah cukup terangkat dan terbaca jelas, tanpa membuat area yang sebelumnya bercahaya menjadi terlalu terang atau kehilangan detail secara berlebihan. Nilai gamma yang lebih besar dari ini cenderung membuat citra terlalu terang dan mulai kehilangan kontras alami pemandangan malam, sedangkan nilai yang lebih kecil belum cukup mengangkat detail di area gelap.
